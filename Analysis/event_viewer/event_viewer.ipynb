# KMC Event Viewer — Jupyter inline mode

This notebook launches the Dash app (`Analysis/event_viewer/app.py`) as an inline
iframe in the next cell. All features work identically to the standalone app:
cascade dropdowns, CNA coloring, sphere rendering, vacancy overlay, atom-radius
slider, projection toggle, and the ← ↑ → frame hotkeys (click into the iframe
once so it has keyboard focus).

Install the package first (`pip install -e .` from the repository root). In the
setup cell, set `RUN_DIR` to view one pyKMC run, or `DATA_ROOT` to browse a folder
of runs laid out as `{composition}/{N}vac/T{temp}/{variant}/`. Leave both as
`None` to use the `PYKMC_ANALYSIS_DATA_ROOT` environment variable.

**Need a bigger view?** Click the **Pop out ↗** button in the top-right of the
embedded viewer — it opens the app in a full browser tab at the same URL. Or
scroll down to the alternative launch cell for `jupyter_mode='tab'`, which
opens a new tab on launch and skips the iframe entirely.

If port **8051** is in use, change `PORT` below. Restart the kernel and re-run
the cells to pick up edits to `plots.py` / `analysis.py` / `app.py`.


````python
# Setup — choose what to open, then import the app.
import os

RUN_DIR = None    # e.g. '/path/to/run'  (one run directory with reference_table.pickle)
DATA_ROOT = None  # e.g. '/path/to/data' (a folder of runs)

if RUN_DIR:
    os.environ['EVENT_VIEWER_RUN'] = str(RUN_DIR)
if DATA_ROOT:
    os.environ['PYKMC_ANALYSIS_DATA_ROOT'] = str(DATA_ROOT)

from Analysis.event_viewer.app import app

PORT = 8051  # change if something else is bound here
````

````python
# Launch inline. Suppress the reloader (it does not play well with Jupyter kernels).
app.run(
    jupyter_mode='inline',
    jupyter_height=900,
    port=PORT,
    debug=False,
    dev_tools_hot_reload=False,
)
````

## Alternative view modes

If you'd rather have the app in a separate tab or as a JupyterLab side panel,
stop the kernel (Interrupt + Restart) and change `jupyter_mode`:

| Value | Shows up as |
|---|---|
| `inline` | iframe inside this cell's output (default above) |
| `external` | prints URL only — open in a separate window |
| `tab` | new browser tab |
| `jupyterlab` | docked panel on the side (JupyterLab only) |

## Quick programmatic access

If you want to use the figure builders directly (skip Dash, produce a static Plotly
figure), import them instead of the app:

```python
from Analysis.sim_discovery import sim_from_dir
from Analysis.event_viewer.data import load_events, load_classifications
from Analysis.event_viewer.plots import make_event_figure

sim = sim_from_dir('/path/to/run')
events = load_events(sim, load_classifications())
row = events.iloc[0]
fig = make_event_figure(row, frame='saddle', render_style='spheres', atom_size=0.9)
fig.show()
```


````python
# Pop-out launch — open directly in a new browser tab (no iframe).
# Interrupt + restart the kernel first if you previously ran the inline cell,
# then run THIS cell instead.
app.run(
    jupyter_mode='tab',
    port=PORT,
    debug=False,
    dev_tools_hot_reload=False,
)
````
